# LeetCode 1581. Customer Who Visited but Did Not Make Any Transactions

Link: https://leetcode.com/problems/customer-who-visited-but-did-not-make-any-transactions/description/?envType=study-plan-v2&envId=top-sql-50

Solution:
```sql
SELECT vs.customer_id, COUNT(vs.visit_id) AS count_no_trans
FROM Visits vs LEFT JOIN Transactions txn ON vs.visit_id = txn.visit_id
WHERE txn.transaction_id is null
GROUP BY vs.customer_id
```


Solution 2:
```sql
SELECT customer_id, COUNT(visit_id) AS count_no_trans 
FROM Visits 
WHERE 
  visit_id NOT IN (
    SELECT visit_id 
    FROM Transactions
  ) 
GROUP BY customer_id
```

## What I learned:
- `LEFT JOIN` keeps columns from both tables, meaning the result will contain 2 `visit_id` columns: one from the `Visits` table and one from the `Transactions` table.
    - E.g.
        ```sql
        SELECT *
        FROM Visits vs LEFT JOIN Transactions txn ON vs.visit_id = txn.visit_id;
        ```
    - This will create a table that contains the following columns:
        - `visit_id` (from `Visits` table)
        - `customer_id` (from `Visits` table)             <-- this column will **always have a value** because it's from the left table in the LEFT JOIN
        - `transaction_id` (from `Transactions` table)       <-- this column will be **NULL** if there is no corresponding transaction
        - `visit_id` (from `Transactions` table)            <-- this column will be **NULL** if there is no corresponding transaction
        - `customer_id` (from `Transactions` table)         <-- this column will be **NULL** if there is no corresponding transaction

- It's hard to visualise the output table after `GROUP BY` because the rows are aggregated, meaning they are collapsed/ summarised into **1 output row per group**
	- E.g., we have:
		| customer_id | visit_id   |
		|-------------|------------|
		| 1           |  10        |
		| 1           |  11        |
		| 2           |  12        |
		| 2           |  13        |
        | 2           |  14        |
    - If we run:
        ```sql
        SELECT *
        FROM visits
        GROUP BY customer_id;
        ```
    - The output will be
        | customer_id | visit_id   |
        |-------------|------------|
        | 1           |  10        |
        | 2           |  12        |

        - This happens because when we use `GROUP BY customer_id`, the database **collapses all rows with the same** `customer_id` **into a single row**. 
        - The `visit_id` shown is typically the first one encountered for that `customer_id`, but it is not guaranteed which one it will be.

- After `GROUP BY`, the aggregate function (like `COUNT`) will be **applied to each group separately** 
    - So, `COUNT(*)` will count the number of rows in each group (which is `customer_id` in this case).
