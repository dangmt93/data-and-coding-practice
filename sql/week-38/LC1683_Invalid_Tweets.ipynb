# LeetCode 1683. Invalid Tweets

Link: https://leetcode.com/problems/invalid-tweets/description/?envType=study-plan-v2&envId=top-sql-50

Solution:
```sql
SELECT *
FROM Tweets
WHERE LENGTH(content) > 15;
```

What I learned:
1. Use `LENGTH(str)` to get the number of characters in a string. This is syntax in MySQL and PostgreSQL
    - For SQL Server, it is `LEN(str)`
1. There is a difference between LENGTH() and CHAR_LENGTH(), but for most cases, `LENGTH()` works to get the number of characters in an English string.
    - `LENGTH()` returns the length of the string in __bytes__.
    - `CHAR_LENGTH()` returns the length of the string in __characters__.
